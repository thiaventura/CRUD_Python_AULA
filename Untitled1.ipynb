```python
import math
import numpy as np
import matplotlib.pyplot as plt
# 0 - Função logaritmica

print("Função logarítmica")
# 1 - Conhecendo a fórmula

print("f(x) = logbx")
# 2 - Construindo a função

while True:
    b = float(input("Digite o valor da base: "))
    if b <= 0 or b == 1:
        print("Digite um valor acima de 0 e diferente de 1")
        print("A base precisa ser um número real e positivo com exceção do 0")
        continue
    else:
        print(f"Beleza carinha! Vamo montando nossa equação: f(x) = log{b}x")
# 3 - Crescente ou Decrescente

        if b > 1:
            print("Já que a base é > 1, então a função é crescente")
        else:
            print("Já que a base é > 0 e < 1, então a função é decrescente")
# 4 - Logaritmando

        x = float(input("Digite o valor do logaritmando: "))
        while x <= 0:
            print("Digite um valor acima de 0")
            print("O logaritmando precisa ser um número real e positivo com exceção do 0")
            x = float(input("Digite o valor do logaritmando: "))
            continue
        else:
            print(f"Muito massa, carinha! Vamo ver como ficou a equação até agora: f({x}) = log{b} {x}")
            fx = math.log(x, b)
            print(f"f({x:.2f}) = {fx:.2f}")
            son = input("Deseja fazer outra função? Digite s/n: ")
            if son == "s":
                continue
            else:
                print("Boa carinha! tua função ta pronta então. Vamo ver como ficou o gráfico: ")
# 5 - Construir o gráfico
                valores_x = np.linspace(0.01, 10, 500)
                valores_y = np.log(valores_x) / np.log(b)

                plt.figure(figsize=(8, 5))
                plt.plot(valores_x, valores_y, label=f"f(x) = log_{b}(x)")

# Destacar o ponto digitado pelo usuário
                plt.scatter(x, fx, color="red", label=f"({x}, {fx:.2f})")

                plt.axhline(y=0, color="black", linewidth=0.8)
                plt.axvline(x=0, color="black", linewidth=0.8)
                plt.grid(True)
                plt.xlabel("x")
                plt.ylabel("f(x)")
                plt.title("Gráfico da função logarítmica")
                plt.legend()
                plt.show()
                break
```
